## 2. Annotating with Your Observations
Good notebooks are well-documented! You can add explanations, observations, and comments to your code to make it understandable for yourself and others.


````python
# ### In Code Cells (using comments)
# In Python code cells, you use the `#` symbol to start a comment. Anything after `#` on that line is ignored by the Python interpreter.

# This is a single-line comment explaining the next line of code.
my_variable = "This is a string."

# You can also add comments at the end of a line of code.
print(my_variable) # This prints the value of my_variable.

# For multi-line comments or explanations that span several lines, you can use triple quotes (''' or """) to create a multi-line string. While technically a string literal, if it's not assigned to a variable, it acts like a multi-line comment.
'''
This is an observation:
I've noticed that Colab is very user-friendly for getting started with Python.
I can easily run code and see the output right here in the notebook.
'''
````

### 2.1 Understanding Cell Order and State
In Jupyter notebooks, the order in which you run cells matters. If you define a variable in one cell, it stays in the 'memory' of the notebook for all other cells, as long as the runtime is active.

Notice the number in the brackets `[ ]` next to a code cell. This indicates the order in which cells were executed.

````python
# Run this cell first
shared_value = "I was defined in an earlier cell!"
````

````python
# Even though this is a different cell, it remembers 'shared_value'
print(shared_value)
````

### 3.1 The File Browser
On the left-hand sidebar, you can click the **folder icon** to see the files currently available in your Colab instance.

*   **sample_data/**: Colab provides some practice datasets here.

````python
# You can also write code to download the *outputs* of your work.

# For instance, this code block reads in the sample data 'california_housing_test.csv',
# Computes some standard statistics,
# and writes the result to file.

# Then it uses some python code that interacts with the Google servers to download it.
import pandas as pd
from google.colab import files

# 1. Read in the sample data
df = pd.read_csv('/content/sample_data/california_housing_test.csv')
#("the variable df contains the result of the command `read_csv` found in the library called pandas, for working with tabular data")

# 2. Compute standard statistics
stats = df.describe()
#("the variable contains the result of the command `describe` computed against the variable df")

# 3. Write the result to a file
file_name = 'housing_stats.csv'
stats.to_csv(file_name)

print(f'Summary statistics saved to {file_name}.')

# 4. Trigger the browser download
files.download(file_name)
````

## 4. Installing Libraries
Colab comes with most data science libraries (like `pandas` or `numpy`) pre-installed. However, if you need a specific library that isn't there, you can install it using a 'bang' (`!`) command.

````python
# The '!' tells Colab to run this as a system command rather than Python code
!pip install wikipedia
````

````python
import wikipedia
import pandas as pd

# 1. Define the search query
query = "Digital Archaeology"
search_results = wikipedia.search(query)

# 2. Collect metadata for each search result
data = []
for title in search_results:
    try:
        page = wikipedia.page(title, auto_suggest=False)
        data.append({
            "Title": page.title,
            "Summary": page.summary[:200] + "...",  # Truncate for readability
            "URL": page.url
        })
    except (wikipedia.exceptions.DisambiguationError, wikipedia.exceptions.PageError):
        continue

# 3. Create a DataFrame
df_wiki = pd.DataFrame(data)

# Display the dataframe
df_wiki
````
